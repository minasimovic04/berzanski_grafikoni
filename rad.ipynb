```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
def ucitaj_tickere(putanja_do_fajla="tickers.txt"):
    if not os.path.exists(putanja_do_fajla):
        print(f"Putanja do fajla {putanja_do_fajla} ne postoji!")
        return []
    with open(putanja_do_fajla,'r') as f:
        tickeri=[]
        for linija in f:
          cista_linija=linija.strip()
          if cista_linija !="":
              velika_slova=cista_linija.upper()
              tickeri.append(velika_slova)

    return tickeri
```

```python
def generisi_izvestaj_za_ticker(ticker,output_dir="izvestaji"):
    try:
        print(f"Preuzimam podatke za {ticker} . . .")
        tk=yf.Ticker(ticker)
        podaci=tk.history(period="6mo")

        if(podaci.empty):
            print(f"Upozorenje: Nema dostupnih podataka za {ticker}! Preskacem..")
            return
        podaci["MA20"]=podaci["Close"].rolling(window=20).mean()
        podaci["MA50"]=podaci["Close"].rolling(window=50).mean()

        plt.figure(figsize=(10,5))
        plt.plot(
            podaci.index,
            podaci["Close"],
            label="Cena zatvaranja",
            color="#1f77b4",
            linewidth=2
        )
        plt.plot(
            podaci.index,
            podaci["MA20"],
            label="20-dnevni prosek (MA20)",
            color="orange",
            linestyle="--"
        )
        plt.plot(
            podaci.index,
            podaci["MA50"],
            label="50-dnevni prosek (MA50)",
            color="gray",
            linestyle=":"
        )

        plt.title(
            f"{ticker} Analiza kretanja cene akcije",
            fontsize=14,
            fontweight='bold'
        )
        plt.xlabel("Datum")
        plt.ylabel("Cena (USD)")
        plt.grid(True,linestyle=':',alpha=0.7)
        plt.legend()
        plt.tight_layout()

        putanja_slike=os.path.join(output_dir,f"{ticker}_grafikon.png")
        plt.savefig(putanja_slike,dpi=300)
        plt.close()
        print(f" Uspešno generisan grafikon za {ticker}")
    except Exception as e:
        print(f" Greška pri obradi tickera {ticker}: {e}")
```

```python
def main():
    izlazni_folder="izvestaji"
    if not os.path.exists(izlazni_folder):
        os.makedirs(izlazni_folder)
    kompanije=ucitaj_tickere("tickers.txt")

    if not kompanije:
        print("Lista je prazna.Proveri fajl tickers.txt")
        return
    print(f"Zapocinjem obradu za {len(kompanije)} kompanija . . .")

    for ticker in kompanije:
        generisi_izvestaj_za_ticker(ticker,izlazni_folder)
    print(
      "\n Proces završen! Svi izveštaji se nalaze u folderu"
      f" '{izlazni_folder}'."
    )
if __name__ == "__main__":
  main()
```
